```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def check_adult(self):
        if self.age > 20 :
          print(self.name)
          print(self.age)

human1 = Human('ichiro', 25)
human2 = Human('jiro', 18)
human3 = Human('saburo', 30)
human1.check_adult()
human2.check_adult()
human3.check_adult()
```
