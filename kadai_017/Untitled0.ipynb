```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def check_adult(self):
    if self.age >= 20 :
      print(f"{self.name}は{self.age}歳で、成人しています。")
    else :
      print(f"{self.name}は{self.age}歳で、未成年です。")

humans = [Human('ichiro', 25),Human('jiro', 18),Human('saburo', 30)]
for human in humans :
  human.check_adult()
```
